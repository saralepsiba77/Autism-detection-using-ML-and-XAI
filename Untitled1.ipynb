```python
# ============================================================
# STEP 4: OUTLIER DETECTION & CONVERSION (FIXED)
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- FIX 1: Replace '?' with NaN and convert age to numeric ---
for df in [child_clean, adolescent_clean, adult_clean]:
    df.replace('?', np.nan, inplace=True)
    df['age'] = pd.to_numeric(df['age'], errors='coerce')

# Columns to check
outlier_cols = ['age', 'result']

# --- Outlier capping function ---
def cap_outliers(df, col):
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    n_out = ((df[col] < lower) | (df[col] > upper)).sum()

    df[col] = np.where(df[col] < lower, lower, df[col])
    df[col] = np.where(df[col] > upper, upper, df[col])

    return df, n_out, lower, upper

# --- Process each dataset ---
datasets = {
    "Child"      : child_clean.copy(),
    "Adolescent" : adolescent_clean.copy(),
    "Adult"      : adult_clean.copy()
}

print("=" * 65)
print("OUTLIER DETECTION & CONVERSION")
print("=" * 65)

for name, df in datasets.items():
    print(f"\n--- {name} ---")
    for col in outlier_cols:
        df, n_out, low, up = cap_outliers(df, col)
        print(f"  {col:<8}: {n_out} outliers capped  (bounds: {low:.1f} to {up:.1f})")
    datasets[name] = df

child_clean      = datasets["Child"]
adolescent_clean = datasets["Adolescent"]
adult_clean      = datasets["Adult"]

print("\n✅ Outliers capped")
```

```python
# ============================================================
# STEP 5 - PART A: COUNT MISSING VALUES (BEFORE KNN)
# ============================================================

import numpy as np
import pandas as pd

# Replace '?' with NaN
for df in [child_clean, adolescent_clean, adult_clean]:
    df.replace('?', np.nan, inplace=True)

print("=" * 60)
print("MISSING VALUES BEFORE KNN")
print("=" * 60)

for name, df in [("Child", child_clean),
                 ("Adolescent", adolescent_clean),
                 ("Adult", adult_clean)]:
    miss = df.isnull().sum()
    miss = miss[miss > 0]
    print(f"\n--- {name} ({df.shape[0]} rows) ---")
    if len(miss) == 0:
        print("No missing values")
    else:
        for col, cnt in miss.items():
            print(f"  {col:<20}: {cnt}")
    print(f"  TOTAL missing : {df.isnull().sum().sum()}")
```

```python
# ============================================================
# STEP 5 - PART B: KNN IMPUTATION
# ============================================================

from sklearn.impute import KNNImputer
from sklearn.preprocessing import LabelEncoder

def knn_impute(df, k=5):
    df = df.copy()

    # --- Separate numeric and categorical columns ---
    num_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    cat_cols = df.select_dtypes(include=['object']).columns.tolist()

    # Remove target from categorical encoding (we don't want to impute target)
    target = 'Class/ASD'
    if target in cat_cols:
        cat_cols.remove(target)

    # --- Encode categorical columns ---
    encoders = {}
    for col in cat_cols:
        le = LabelEncoder()
        # Fill NaN temporarily with a placeholder so LabelEncoder works
        non_null = df[col].dropna().astype(str)
        le.fit(non_null)
        encoders[col] = le

        # Encode (NaN stays NaN)
        df[col] = df[col].apply(
            lambda x: le.transform([str(x)])[0] if pd.notnull(x) else np.nan
        )

    # --- Apply KNN Imputer on all feature columns (numeric + encoded categorical) ---
    feature_cols = num_cols + cat_cols
    imputer = KNNImputer(n_neighbors=k)

    df_imputed = df.copy()
    df_imputed[feature_cols] = imputer.fit_transform(df[feature_cols])

    # --- Decode categorical columns back to original labels ---
    for col in cat_cols:
        le = encoders[col]
        df_imputed[col] = df_imputed[col].round().astype(int)
        df_imputed[col] = df_imputed[col].apply(
            lambda x: le.inverse_transform([x])[0] if 0 <= x < len(le.classes_) else np.nan
        )

    return df_imputed

# Apply KNN on each dataset
child_clean      = knn_impute(child_clean, k=5)
adolescent_clean = knn_impute(adolescent_clean, k=5)
adult_clean      = knn_impute(adult_clean, k=5)

print("✅ KNN imputation done")
```

```python
# ============================================================
# TVAE AUGMENTATION — REPRODUCIBLE (SEED + CSV SAVE)
# ============================================================
# First run  : trains TVAE, generates, saves CSVs to tvae_output/
# Later runs : loads the CSVs directly → identical data, no retraining
# ============================================================

import os
import random
import numpy as np
import pandas as pd
import torch
from ctgan import TVAE

# ---------- 1. SET ALL SEEDS ----------
SEED = 42

def set_seed(seed=SEED):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
    os.environ["PYTHONHASHSEED"] = str(seed)

set_seed()

# ---------- 2. OUTPUT DIRECTORY ----------
SAVE_DIR = "tvae_output"
os.makedirs(SAVE_DIR, exist_ok=True)

files = {
    n: f"{SAVE_DIR}/{n}_augmented.csv"
    for n in ["Child", "Adolescent", "Adult"]
}

# ---------- 3. COLUMN GROUPS ----------
score_cols = [f"A{i}_Score" for i in range(1, 11)]
cat_cols = [
    "gender", "ethnicity", "jundice", "austim",
    "contry_of_res", "used_app_before", "relation"
]
target = "Class/ASD"

# ---------- 4. IF FILES EXIST → LOAD, DON'T REGENERATE ----------
if all(os.path.exists(p) for p in files.values()):
    final = {n: pd.read_csv(p) for n, p in files.items()}
    print("✅ Loaded saved TVAE data (nothing was regenerated)")
    print("=" * 60)
    for n, df in final.items():
        print(f"  {n:<12}: {len(df)} rows")
    print("=" * 60)

# ---------- 5. OTHERWISE → GENERATE ONCE AND SAVE ----------
else:
    real = {
        "Child":      child_clean.drop_duplicates().reset_index(drop=True),
        "Adolescent": adolescent_clean.drop_duplicates().reset_index(drop=True),
        "Adult":      adult_clean.drop_duplicates().reset_index(drop=True),
    }
    aug_ratio = {"Child": 1.5, "Adolescent": 2.0, "Adult": 1.5}
    discrete_cols = cat_cols + score_cols + [target] + ["age", "result"]

    # --- Cleaning function for synthetic data ---
    def clean_synthetic(s, real_df):
        s = s.copy()
        for c in ["age"] + score_cols:
            s[c] = pd.to_numeric(s[c], errors="coerce")
        s = s.dropna().copy()
        s["age"] = s["age"].round().clip(
            real_df["age"].min(), real_df["age"].max()
        ).astype(int)
        for c in score_cols:
            s[c] = s[c].round().astype(int)
        s["result"] = s[score_cols].sum(axis=1)
        return s[real_df.columns]

    # --- Show plan ---
    print("=" * 65)
    print("TVAE AUGMENTATION PLAN (FIRST RUN — will save CSVs)")
    print("=" * 65)
    total_real = total_synth = 0
    for name, df in real.items():
        n_real = len(df)
        n_synth = int(n_real * aug_ratio[name])
        total_real += n_real
        total_synth += n_synth
        print(f"{name:<12}: real = {n_real:>4} | ratio = {aug_ratio[name]} "
              f"| synthetic = {n_synth:>4} | total = {n_real + n_synth:>4}")
    print("-" * 65)
    print(f"{'TOTAL':<12}: real = {total_real:>4} | "
          f"synthetic = {total_synth:>4} | "
          f"total = {total_real + total_synth:>4}")
    print("=" * 65)

    # --- Train + generate per dataset ---
    final = {}
    for name, df in real.items():
        set_seed()  # reset seed per dataset
        print(f"\nTraining TVAE on {name} ({len(df)} rows)...")
        model = TVAE(epochs=300, batch_size=100)

        # Seed TVAE's internal generator if supported
        if hasattr(model, "set_random_state"):
            model.set_random_state(SEED)

        model.fit(df, discrete_columns=discrete_cols)

        need = int(len(df) * aug_ratio[name])
        parts, collected, tries = [], pd.DataFrame(columns=df.columns), 0

        while len(collected) < need and tries < 10:
            batch = clean_synthetic(model.sample(need * 2), df).drop_duplicates()
            merged = batch.merge(df, how="left", indicator=True)
            batch = merged[merged["_merge"] == "left_only"].drop(columns="_merge")
            parts.append(batch)
            collected = pd.concat(parts).drop_duplicates()
            tries += 1

        synth = collected.head(need).reset_index(drop=True)
        final[name] = pd.concat([df, synth], ignore_index=True)

        # Save CSV
        final[name].to_csv(files[name], index=False)

        print(f"  {name:<12}: real = {len(df):>4} | "
              f"synthetic = {len(synth):>4} | "
              f"saved to {files[name]}")

    print("\n" + "=" * 65)
    print("✅ Generated ONCE and saved to tvae_output/")
    print("   Next runs will LOAD these CSVs → identical results.")
    print("=" * 65)

# ---------- 6. SUMMARY ----------
print("\nFINAL DATASET SIZES:")
print("-" * 50)
grand_total = 0
for name, df in final.items():
    grand_total += len(df)
    print(f"  {name:<12}: {len(df):>5} rows")
print("-" * 50)
print(f"  {'GRAND TOTAL':<12}: {grand_total:>5} rows")
```

```python
# ============================================================
# CELL 1: ENCODE CATEGORICAL COLUMNS FOR SMOTE-TOMEK
# ============================================================

import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder

# Categorical columns to encode
cat_cols_encode = ["gender", "ethnicity", "jundice", "austim",
                   "contry_of_res", "used_app_before", "relation"]

# Encode each dataset separately, then SMOTE-Tomek separately
encoded_final = {}
encoders_final = {}

for name, df in final.items():
    df_enc = df.copy()
    encoders_final[name] = {}

    for col in cat_cols_encode:
        le = LabelEncoder()
        df_enc[col] = le.fit_transform(df_enc[col].astype(str))
        encoders_final[name][col] = le

    # Encode target: NO=0, YES=1
    df_enc["Class/ASD"] = (df_enc["Class/ASD"] == "YES").astype(int)

    encoded_final[name] = df_enc
    print(f"{name:<12}: shape = {df_enc.shape} | "
          f"YES = {(df_enc['Class/ASD']==1).sum()} | "
          f"NO = {(df_enc['Class/ASD']==0).sum()}")

print("\n✅ Encoding done")
```

```python
# ============================================================
# CELL 2: SMOTE-TOMEK
# ============================================================

from imblearn.combine import SMOTETomek

balanced = {}

for name, df_enc in encoded_final.items():
    print(f"\nApplying SMOTE-Tomek on {name}...")

    X = df_enc.drop(columns=["Class/ASD"])
    y = df_enc["Class/ASD"]

    print(f"  Before: YES = {(y==1).sum()}, NO = {(y==0).sum()}")

    smt = SMOTETomek(random_state=42)
    X_res, y_res = smt.fit_resample(X, y)

    # Combine back
    df_bal = pd.concat([X_res, y_res], axis=1)
    balanced[name] = df_bal

    print(f"  After : YES = {(y_res==1).sum()}, NO = {(y_res==0).sum()}")
    print(f"  Total : {len(df_bal)} rows")

print("\n✅ SMOTE-Tomek done")
```

```python
# ============================================================
# CELL 2: FREEZE TEST SET (prevents accidental modification)
# ============================================================

# Create a "locked" snapshot of the test set
locked_test = {}

for name, s in splits.items():
    locked_test[name] = {
        "X_test": s["X_test"].copy(),
        "y_test": s["y_test"].copy(),
    }

print("=" * 60)
print("LOCKED TEST SETS")
print("=" * 60)
for name, s in locked_test.items():
    print(f"{name:<12}: X_test = {s['X_test'].shape}, "
          f"y_test = {s['y_test'].shape}")

print("\n🔒 Test sets are now frozen.")
print("   Train only on `splits[name]['X_train']` and "
      "`splits[name]['y_train']`.")
print("   Evaluate ONLY on `locked_test[name]` at the end.")
```

```python


# ============================================================
# RETRAIN META MODEL — Proper Stacking with OOF Predictions
# ============================================================

import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedKFold
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                              f1_score, roc_auc_score, confusion_matrix)

RANDOM_STATE = 42
N_SPLITS     = 5

meta_results_v2 = {}
meta_models_v2  = {}
oof_meta_data   = {}   # store OOF features for inspection

print("=" * 70)
print("RETRAINING META MODEL (Out-of-Fold Stacking)")
print("=" * 70)

for name in splits.keys():
    X_train = splits[name]["X_train"]
    y_train = splits[name]["y_train"]
    X_test  = locked_test[name]["X_test"]
    y_test  = locked_test[name]["y_test"]

    n_train = len(X_train)

    # --------------------------------------------------------
    # 1. Build OOF meta-features on TRAIN set
    # --------------------------------------------------------
    oof = np.zeros((n_train, 3))   # [RF, SVM, NB]

    skf = StratifiedKFold(n_splits=N_SPLITS, shuffle=True,
                          random_state=RANDOM_STATE)

    for fold, (tr_idx, val_idx) in enumerate(skf.split(X_train, y_train)):
        X_tr, X_val = X_train.iloc[tr_idx], X_train.iloc[val_idx]
        y_tr        = y_train.iloc[tr_idx]

        # --- RF ---
        rf_f = RandomForestClassifier(n_estimators=200, random_state=RANDOM_STATE,
                                      n_jobs=-1)
        rf_f.fit(X_tr, y_tr)
        oof[val_idx, 0] = rf_f.predict_proba(X_val)[:, 1]

        # --- SVM (scaled) ---
        sc_f = StandardScaler().fit(X_tr)
        svm_f = SVC(kernel="rbf", C=1.0, gamma="scale",
                    probability=True, random_state=RANDOM_STATE)
        svm_f.fit(sc_f.transform(X_tr), y_tr)
        oof[val_idx, 1] = svm_f.predict_proba(sc_f.transform(X_val))[:, 1]

        # --- NB (scaled) ---
        nb_f = GaussianNB()
        nb_f.fit(sc_f.transform(X_tr), y_tr)
        oof[val_idx, 2] = nb_f.predict_proba(sc_f.transform(X_val))[:, 1]

    oof_meta_data[name] = oof

    # --------------------------------------------------------
    # 2. Train meta-learner on OOF features
    # --------------------------------------------------------
    meta = LogisticRegression(random_state=RANDOM_STATE, max_iter=1000)
    meta.fit(oof, y_train)

    # --------------------------------------------------------
    # 3. Build TEST meta-features using the EXISTING base models
    #    (rf_models / svm_models / nb_models were already fit
    #     on the full X_train in earlier cells)
    # --------------------------------------------------------
    rf_test_prob  = rf_models[name].predict_proba(X_test)[:, 1]
    svm_test_prob = svm_models[name].predict_proba(
                        svm_scalers[name].transform(X_test))[:, 1]
    nb_test_prob  = nb_models[name].predict_proba(
                        nb_scalers[name].transform(X_test))[:, 1]

    X_meta_test = np.column_stack([rf_test_prob, svm_test_prob, nb_test_prob])

    # --------------------------------------------------------
    # 4. Predict & evaluate on locked test set
    # --------------------------------------------------------
    y_pred = meta.predict(X_meta_test)
    y_prob = meta.predict_proba(X_meta_test)[:, 1]

    acc  = accuracy_score(y_test, y_pred)
    prec = precision_score(y_test, y_pred, zero_division=0)
    rec  = recall_score(y_test, y_pred, zero_division=0)
    f1   = f1_score(y_test, y_pred, zero_division=0)
    auc  = roc_auc_score(y_test, y_prob)
    cm   = confusion_matrix(y_test, y_pred)

    meta_results_v2[name] = {
        "accuracy": acc, "precision": prec,
        "recall": rec, "f1": f1, "auc": auc,
        "y_pred": y_pred, "y_prob": y_prob
    }
    meta_models_v2[name] = meta

    print(f"\n--- {name} ---")
    print(f"  Accuracy  : {acc:.4f}")
    print(f"  Precision : {prec:.4f}")
    print(f"  Recall    : {rec:.4f}")
    print(f"  F1-Score  : {f1:.4f}")
    print(f"  ROC-AUC   : {auc:.4f}")
    print(f"  Confusion Matrix:")
    print(f"    TN={cm[0,0]:>4}  FP={cm[0,1]:>4}")
    print(f"    FN={cm[1,0]:>4}  TP={cm[1,1]:>4}")
    print(f"  Meta weights (OOF): RF={meta.coef_[0][0]:+.3f}  "
          f"SVM={meta.coef_[0][1]:+.3f}  NB={meta.coef_[0][2]:+.3f}")

print("\n✅ Meta model (OOF version) retrained for all 3 datasets")
```

```python
# ============================================================
# 1D TABULAR RESNET & DENSENET → RF/SVM/NB/ML
# NO torchvision required — pure PyTorch
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, TensorDataset

from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                              f1_score, roc_auc_score, confusion_matrix)

# Reproducibility
SEED = 42
torch.manual_seed(SEED)
np.random.seed(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")
print("✅ Imports ready (no torchvision needed)")
```

```python
# ============================================================
# 1D RESNET ENCODER (Tabular)
# ============================================================

class ResidualBlock1D(nn.Module):
    def __init__(self, channels):
        super().__init__()
        self.conv1 = nn.Conv1d(channels, channels, kernel_size=3, padding=1)
        self.bn1   = nn.BatchNorm1d(channels)
        self.relu  = nn.ReLU(inplace=True)
        self.conv2 = nn.Conv1d(channels, channels, kernel_size=3, padding=1)
        self.bn2   = nn.BatchNorm1d(channels)

    def forward(self, x):
        residual = x
        out = self.relu(self.bn1(self.conv1(x)))
        out = self.bn2(self.conv2(out))
        out = out + residual
        return self.relu(out)


class TabularResNetEncoder(nn.Module):
    """1D ResNet encoder: extracts feature vector from 1D tabular row."""
    def __init__(self, input_dim=19, hidden_channels=64):
        super().__init__()
        self.start_conv = nn.Conv1d(1, hidden_channels, kernel_size=3, padding=1)
        self.bn   = nn.BatchNorm1d(hidden_channels)
        self.relu = nn.ReLU(inplace=True)
        self.layer1 = ResidualBlock1D(hidden_channels)
        self.layer2 = ResidualBlock1D(hidden_channels)
        self.pool   = nn.AdaptiveAvgPool1d(1)

    def forward(self, x):
        # x: (batch, features) → (batch, 1, features)
        x = x.unsqueeze(1)
        x = self.relu(self.bn(self.start_conv(x)))
        x = self.layer1(x)
        x = self.layer2(x)
        x = self.pool(x)
        return x.view(x.size(0), -1)   # (batch, hidden_channels)


class ResNetClassifier(nn.Module):
    """ResNet encoder + classification head (for training only)."""
    def __init__(self, input_dim, num_classes=2):
        super().__init__()
        self.encoder = TabularResNetEncoder(input_dim)
        self.head = nn.Linear(64, num_classes)

    def forward(self, x):
        return self.head(self.encoder(x))


print("✅ 1D ResNet architecture defined")
```

```python
# ============================================================
# 1D DENSENET ENCODER (Tabular)
# ============================================================

class DenseBlock1D(nn.Module):
    """One dense layer: computes new features, concatenates with input."""
    def __init__(self, in_ch, growth_rate):
        super().__init__()
        self.bn   = nn.BatchNorm1d(in_ch)
        self.relu = nn.ReLU(inplace=True)
        self.conv = nn.Conv1d(in_ch, growth_rate, kernel_size=3, padding=1)

    def forward(self, x):
        new = self.conv(self.relu(self.bn(x)))
        return torch.cat([x, new], dim=1)


class TabularDenseNetEncoder(nn.Module):
    """1D DenseNet encoder: dense connectivity for tabular data."""
    def __init__(self, input_dim=19, init_ch=32, growth=16):
        super().__init__()
        self.start_conv = nn.Conv1d(1, init_ch, kernel_size=3, padding=1)
        self.bn   = nn.BatchNorm1d(init_ch)
        self.relu = nn.ReLU(inplace=True)

        self.dense1 = DenseBlock1D(init_ch,          growth)
        self.dense2 = DenseBlock1D(init_ch + growth, growth)
        self.dense3 = DenseBlock1D(init_ch + 2*growth, growth)

        self.pool = nn.AdaptiveAvgPool1d(1)
        self.out_ch = init_ch + 3 * growth

    def forward(self, x):
        x = x.unsqueeze(1)
        x = self.relu(self.bn(self.start_conv(x)))
        x = self.dense1(x)
        x = self.dense2(x)
        x = self.dense3(x)
        x = self.pool(x)
        return x.view(x.size(0), -1)   # (batch, out_ch)


class DenseNetClassifier(nn.Module):
    def __init__(self, input_dim, num_classes=2):
        super().__init__()
        self.encoder = TabularDenseNetEncoder(input_dim)
        self.head = nn.Linear(self.encoder.out_ch, num_classes)

    def forward(self, x):
        return self.head(self.encoder(x))


print("✅ 1D DenseNet architecture defined")
```

```python
# ============================================================
# TRAINING FUNCTION — trains encoder + head jointly
# ============================================================

def train_encoder(model, X_train, y_train, X_val, y_val,
                  epochs=40, batch_size=64, lr=1e-3, verbose=False):
    """
    Train a deep classifier (encoder + head).
    Returns the trained model.
    """
    X_tr_t = torch.tensor(X_train.values if hasattr(X_train, "values") else X_train,
                          dtype=torch.float32)
    y_tr_t = torch.tensor(y_train.values if hasattr(y_train, "values") else y_train,
                          dtype=torch.long)
    X_va_t = torch.tensor(X_val.values if hasattr(X_val, "values") else X_val,
                          dtype=torch.float32)
    y_va_t = torch.tensor(y_val.values if hasattr(y_val, "values") else y_val,
                          dtype=torch.long)

    train_ds = TensorDataset(X_tr_t, y_tr_t)
    train_loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True)

    model = model.to(device)
    optimizer = optim.Adam(model.parameters(), lr=lr)
    criterion = nn.CrossEntropyLoss()

    for epoch in range(epochs):
        model.train()
        total_loss = 0
        for xb, yb in train_loader:
            xb, yb = xb.to(device), yb.to(device)
            optimizer.zero_grad()
            logits = model(xb)
            loss = criterion(logits, yb)
            loss.backward()
            optimizer.step()
            total_loss += loss.item()

        # Validation accuracy
        model.eval()
        with torch.no_grad():
            val_logits = model(X_va_t.to(device))
            val_pred = val_logits.argmax(1).cpu().numpy()
            val_acc = (val_pred == y_va_t.numpy()).mean()

        if verbose and (epoch + 1) % 10 == 0:
            print(f"    Epoch {epoch+1:>2}/{epochs} | "
                  f"Loss={total_loss/len(train_loader):.4f} | "
                  f"Val Acc={val_acc:.4f}")

    return model


def extract_features(encoder, X):
    """Extract features from trained encoder (no head)."""
    encoder.eval()
    X_t = torch.tensor(X.values if hasattr(X, "values") else X, dtype=torch.float32).to(device)
    with torch.no_grad():
        feats = encoder(X_t).cpu().numpy()
    return feats


print("✅ Training functions ready")
```

```python
# ============================================================
# FULL PIPELINE: ResNet & DenseNet → RF/SVM/NB/ML
# ============================================================

# Classifiers to test
CLASSIFIERS = {
    "RF":  RandomForestClassifier(n_estimators=200, random_state=SEED, n_jobs=-1),
    "SVM": SVC(kernel="rbf", probability=True, random_state=SEED),
    "NB":  GaussianNB(),
    "ML":  LogisticRegression(max_iter=1000, random_state=SEED),
}

# Results storage
resnet_results_1d   = {}
densenet_results_1d = {}
resnet_cms_1d       = {}
densenet_cms_1d     = {}

print("=" * 70)
print("1D RESNET & DENSENET → ML PIPELINE")
print("=" * 70)

for name in splits.keys():
    print(f"\n{'='*70}")
    print(f"DATASET: {name}")
    print(f"{'='*70}")

    X_train = splits[name]["X_train"]
    X_test  = locked_test[name]["X_test"]
    y_train = splits[name]["y_train"]
    y_test  = locked_test[name]["y_test"]

    input_dim = X_train.shape[1]
    print(f"  Input features: {input_dim}")
    print(f"  Train size: {len(X_train)} | Test size: {len(X_test)}")

    # ---------------- TRAIN RESNET ENCODER ----------------
    print(f"\n  [ResNet] Training encoder (40 epochs)...")
    resnet_model = ResNetClassifier(input_dim=input_dim, num_classes=2)
    resnet_model = train_encoder(
        resnet_model, X_train, y_train, X_test, y_test,
        epochs=40, batch_size=64, lr=1e-3, verbose=True
    )

    # Extract features
    F_train_res = extract_features(resnet_model.encoder, X_train)
    F_test_res  = extract_features(resnet_model.encoder, X_test)
    print(f"  [ResNet] Feature shape: {F_train_res.shape}")

    # ---------------- TRAIN DENSENET ENCODER ----------------
    print(f"\n  [DenseNet] Training encoder (40 epochs)...")
    densenet_model = DenseNetClassifier(input_dim=input_dim, num_classes=2)
    densenet_model = train_encoder(
        densenet_model, X_train, y_train, X_test, y_test,
        epochs=40, batch_size=64, lr=1e-3, verbose=True
    )

    F_train_dense = extract_features(densenet_model.encoder, X_train)
    F_test_dense  = extract_features(densenet_model.encoder, X_test)
    print(f"  [DenseNet] Feature shape: {F_train_dense.shape}")

    # ---------------- RUN 4 ML MODELS ON EACH BRANCH ----------------
    resnet_results_1d[name]   = {}
    densenet_results_1d[name] = {}
    resnet_cms_1d[name]       = {}
    densenet_cms_1d[name]     = {}

    for branch_name, F_tr, F_te in [
        ("ResNet",   F_train_res,   F_test_res),
        ("DenseNet", F_train_dense, F_test_dense),
    ]:
        # Scale for SVM / NB / ML
        scaler = StandardScaler().fit(F_tr)
        F_tr_s = scaler.transform(F_tr)
        F_te_s = scaler.transform(F_te)

        print(f"\n  [{branch_name} branch]")
        for clf_name, clf in CLASSIFIERS.items():
            # RF doesn't need scaled features (but scaling doesn't hurt)
            if clf_name == "RF":
                clf.fit(F_tr, y_train)
                y_pred = clf.predict(F_te)
                y_prob = clf.predict_proba(F_te)[:, 1]
            else:
                clf.fit(F_tr_s, y_train)
                y_pred = clf.predict(F_te_s)
                y_prob = clf.predict_proba(F_te_s)[:, 1]

            acc  = accuracy_score(y_test, y_pred)
            prec = precision_score(y_test, y_pred, zero_division=0)
            rec  = recall_score(y_test, y_pred, zero_division=0)
            f1   = f1_score(y_test, y_pred, zero_division=0)
            auc  = roc_auc_score(y_test, y_prob)
            cm   = confusion_matrix(y_test, y_pred)

            target_dict = resnet_results_1d if branch_name == "ResNet" else densenet_results_1d
            cm_dict     = resnet_cms_1d     if branch_name == "ResNet" else densenet_cms_1d

            target_dict[name][clf_name] = {
                "acc": acc, "prec": prec, "rec": rec,
                "f1": f1, "auc": auc,
                "y_pred": y_pred, "y_prob": y_prob,
            }
            cm_dict[name][clf_name] = cm

            print(f"    {clf_name:<4}: Acc={acc:.4f} | F1={f1:.4f} | AUC={auc:.4f}")

print("\n✅ Full 1D pipeline complete")
```

```python
# ============================================================
# LIME — Local Interpretable Model-agnostic Explanations
# (install once if needed: pip install lime)
# ============================================================

try:
    import lime
    import lime.lime_tabular
except ImportError:
    import subprocess, sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "lime", "-q"])
    import lime
    import lime.lime_tabular

import matplotlib.pyplot as plt

# LIME works best on a model you can hand a predict_proba callable
# We'll explain the best model per dataset.
model_objects = {
    "Random Forest": rf_models,
    "SVM":           svm_models,
    "Naive Bayes":   nb_models,
    "Meta (Stack)":  meta_models_v2,
}

for name in splits.keys():
    print(f"\n{'='*60}\nLIME for {name} (best model = {best_per_dataset[name]})\n{'='*60}")

    X_train = splits[name]["X_train"]
    X_test  = locked_test[name]["X_test"]
    y_test  = locked_test[name]["y_test"]

    best_model_name = best_per_dataset[name]

    # Build a predict_proba-compatible callable
    if best_model_name == "Random Forest":
        model = rf_models[name]
        predict_fn = model.predict_proba
    elif best_model_name == "SVM":
        model = svm_models[name]
        scaler = svm_scalers[name]
        predict_fn = lambda x: model.predict_proba(scaler.transform(x))
    elif best_model_name == "Naive Bayes":
        model = nb_models[name]
        scaler = nb_scalers[name]
        predict_fn = lambda x: model.predict_proba(scaler.transform(x))
    else:  # Meta
        # Meta takes [rf_prob, svm_prob, nb_prob] — build wrapper
        meta = meta_models_v2[name]
        rf_m   = rf_models[name]
        svm_m  = svm_models[name]; svm_s = svm_scalers[name]
        nb_m   = nb_models[name];  nb_s  = nb_scalers[name]
        def predict_fn(x):
            p_rf  = rf_m.predict_proba(x)[:, 1]
            p_svm = svm_m.predict_proba(svm_s.transform(x))[:, 1]
            p_nb  = nb_m.predict_proba(nb_s.transform(x))[:, 1]
            meta_in = np.column_stack([p_rf, p_svm, p_nb])
            return meta.predict_proba(meta_in)

    # LIME explainer
    explainer_lime = lime.lime_tabular.LimeTabularExplainer(
        training_data=X_train.values,
        feature_names=list(X_train.columns),
        class_names=["NO", "YES"],
        mode="classification",
        discretize_continuous=True,
        random_state=42,
    )

    # Explain one YES prediction (correctly classified)
    y_pred = model_results_all[best_model_name][name]["y_pred"]
    candidates = np.where((y_test.values == 1) & (y_pred == 1))[0]
    if len(candidates) == 0:
        candidates = np.arange(len(y_test))
    sample_idx = int(candidates[0])

    exp = explainer_lime.explain_instance(
        X_test.iloc[sample_idx].values,
        predict_fn,
        num_features=10,
        top_labels=1,
    )

    # Save as HTML (interactive) and as a matplotlib figure
    exp.save_to_file(f"lime_{name}.html")

    fig = exp.as_pyplot_figure(label=exp.available_labels()[0])
    fig.suptitle(f"{name} — LIME explanation (sample #{sample_idx}, "
                 f"best model = {best_model_name})",
                 fontweight="bold")
    plt.tight_layout()
    plt.savefig(f"lime_{name}.png", dpi=150, bbox_inches="tight")
    plt.show()

    # Print textual explanation
    print(f"  Sample #{sample_idx}: actual=YES, "
          f"pred=YES, prob={model_results_all[best_model_name][name]['y_prob'][sample_idx]:.3f}")
    print("  Top contributing features (LIME):")
    for feat, weight in exp.as_list(label=exp.available_labels()[0])[:10]:
        print(f"    {feat:<45} {weight:+.4f}")

print("\n✅ LIME done — HTML + PNG files saved")
```

```python
# ============================================================
# SHAP — TreeExplainer for Random Forest
# Datasets: Adolescent, Adult
# ============================================================

import shap
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

shap.initjs()

shap_results = {}

for name in ["Adolescent", "Adult"]:
    print("\n" + "=" * 60)
    print(f"SHAP (TreeExplainer) — {name} : Random Forest")
    print("=" * 60)

    X_train = splits[name]["X_train"]
    X_test  = locked_test[name]["X_test"]
    model   = rf_models[name]

    # --- 1. TreeExplainer ---
    explainer   = shap.TreeExplainer(model)
    shap_values = explainer.shap_values(X_test)

    # --- 2. Normalize to 2-D (n_samples, n_features) for YES class ---
    if isinstance(shap_values, list):
        sv_pos = np.asarray(shap_values[1])
    elif np.asarray(shap_values).ndim == 3:
        sv_pos = np.asarray(shap_values)[..., 1]
    else:
        sv_pos = np.asarray(shap_values)

    print(f"[{name}] sv_pos.shape = {sv_pos.shape}")

    # --- 3. Scalar base_value (class 1) ---
    ev = explainer.expected_value
    if isinstance(ev, (list, np.ndarray)):
        ev_arr = np.asarray(ev).ravel()
        base_val = float(ev_arr[1]) if len(ev_arr) > 1 else float(ev_arr[0])
    else:
        base_val = float(ev)

    shap_results[name] = {
        "explainer":   explainer,
        "shap_values": sv_pos,
        "X_test":      X_test,
        "base_value":  base_val,
    }

    # --- 4. Summary (beeswarm) ---
    plt.figure()
    shap.summary_plot(sv_pos, X_test, show=False, plot_type="dot", max_display=15)
    plt.title(f"{name} — SHAP Summary (RF)", fontsize=12, fontweight="bold")
    plt.tight_layout()
    plt.savefig(f"shap_summary_{name.lower()}.png", dpi=150, bbox_inches="tight")
    plt.show()

    # --- 5. Bar plot (mean |SHAP|) ---
    plt.figure()
    shap.summary_plot(sv_pos, X_test, show=False, plot_type="bar", max_display=15)
    plt.title(f"{name} — Mean |SHAP| (RF)", fontsize=12, fontweight="bold")
    plt.tight_layout()
    plt.savefig(f"shap_bar_{name.lower()}.png", dpi=150, bbox_inches="tight")
    plt.show()

    # --- 6. Waterfall on a correctly-predicted YES patient ---
    y_test = locked_test[name]["y_test"].values
    y_pred = rf_results[name]["y_pred"]
    tp_idx = np.where((y_test == 1) & (y_pred == 1))[0]
    sample_idx = tp_idx[0] if len(tp_idx) > 0 else 0

    # Force 1-D row for waterfall
    sv_row = np.asarray(sv_pos[sample_idx]).ravel()

    print(f"[{name}] Waterfall sv_row.shape = {sv_row.shape}  "
          f"base = {base_val:.4f}")
    assert sv_row.shape == (X_test.shape[1],)

    explanation = shap.Explanation(
        values       = sv_row,
        base_values  = base_val,
        data         = X_test.iloc[sample_idx].values.ravel(),
        feature_names= list(X_test.columns),
    )

    plt.figure()
    shap.plots.waterfall(explanation, max_display=12, show=False)
    plt.title(f"{name} — SHAP Waterfall (correct YES sample #{sample_idx})",
              fontsize=12, fontweight="bold")
    plt.tight_layout()
    plt.savefig(f"shap_waterfall_{name.lower()}.png", dpi=150, bbox_inches="tight")
    plt.show()

    print(f"✅ SHAP plots saved for {name}")

print("\n✅ Finished Adolescent and Adult SHAP analysis.")
```

```python
# ============================================================
# SHAP — Child dataset (dual explanation)
#   1. TreeExplainer  → Random Forest       (feature-level)
#   2. KernelExplainer → Meta-Model         (meta-level reasoning)
# ============================================================

import shap
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time
import tracemalloc

name    = "Child"
X_train = splits[name]["X_train"]
X_test  = locked_test[name]["X_test"]
y_test  = locked_test[name]["y_test"]

# ------------------------------------------------------------
# Small helpers — safe across old & new SHAP versions
# ------------------------------------------------------------
def shap_to_2d(shap_values, positive_class=1):
    """Return (n_samples, n_features) SHAP array for the positive class."""
    if isinstance(shap_values, list):
        return np.asarray(shap_values[positive_class])
    arr = np.asarray(shap_values)
    if arr.ndim == 3:
        return arr[..., positive_class]
    return arr

def shap_base_to_float(expected_value, positive_class=1):
    """Scalar base value for positive class."""
    if isinstance(expected_value, (list, np.ndarray)):
        ev = np.asarray(expected_value).ravel()
        return float(ev[positive_class]) if len(ev) > positive_class else float(ev[0])
    return float(expected_value)


# ============================================================
# PART A — TreeExplainer on the underlying Random Forest
# ============================================================
print("=" * 60)
print(f"SHAP (TreeExplainer) — {name} : Random Forest")
print("=" * 60)

rf_explainer = shap.TreeExplainer(rf_models[name])
sv_rf_raw    = rf_explainer.shap_values(X_test)

# --- Normalize to 2-D ---
sv_rf_pos = shap_to_2d(sv_rf_raw, positive_class=1)
base_val_rf = shap_base_to_float(rf_explainer.expected_value, positive_class=1)

print(f"[{name}] RF SHAP array shape = {sv_rf_pos.shape}")
assert sv_rf_pos.ndim == 2 and sv_rf_pos.shape[1] == X_test.shape[1]

shap_results.setdefault(name, {})
shap_results[name].update({
    "explainer_rf":    rf_explainer,
    "shap_values_rf":  sv_rf_pos,
    "X_test":          X_test,
    "base_value_rf":   base_val_rf,
})

# --- Summary plot (beeswarm) ---
plt.figure()
shap.summary_plot(sv_rf_pos, X_test, show=False, plot_type="dot", max_display=15)
plt.title(f"{name} — SHAP Summary (Random Forest)", fontsize=12, fontweight="bold")
plt.tight_layout()
plt.savefig(f"shap_summary_{name.lower()}_rf.png", dpi=150, bbox_inches="tight")
plt.show()

# --- Bar plot (mean |SHAP|) ---
plt.figure()
shap.summary_plot(sv_rf_pos, X_test, show=False, plot_type="bar", max_display=15)
plt.title(f"{name} — Mean |SHAP| (Random Forest)", fontsize=12, fontweight="bold")
plt.tight_layout()
plt.savefig(f"shap_bar_{name.lower()}_rf.png", dpi=150, bbox_inches="tight")
plt.show()

# --- Waterfall on a correct YES sample ---
y_pred_rf  = rf_results[name]["y_pred"]
tp_idx     = np.where((y_test.values == 1) & (y_pred_rf == 1))[0]
sample_idx = tp_idx[0] if len(tp_idx) > 0 else 0

sv_row = np.asarray(sv_rf_pos[sample_idx]).ravel()   # force 1-D
print(f"[{name}] Waterfall sv_row.shape = {sv_row.shape} | base = {base_val_rf:.4f}")
assert sv_row.shape == (X_test.shape[1],)

explanation_rf = shap.Explanation(
    values        = sv_row,
    base_values   = base_val_rf,
    data          = X_test.iloc[sample_idx].values.ravel(),
    feature_names = list(X_test.columns),
)

plt.figure()
shap.plots.waterfall(explanation_rf, max_display=12, show=False)
plt.title(f"{name} — SHAP Waterfall (RF, correct YES sample #{sample_idx})",
          fontsize=12, fontweight="bold")
plt.tight_layout()
plt.savefig(f"shap_waterfall_{name.lower()}_rf.png", dpi=150, bbox_inches="tight")
plt.show()


# ============================================================
# PART B — KernelExplainer on the Meta-Model
# ============================================================
print("\n" + "=" * 60)
print(f"SHAP (KernelExplainer) — {name} : Meta-Model")
print("=" * 60)

# --- Build meta-features (RF / SVM / NB probabilities) ---
def build_meta_features(X):
    rf_p  = rf_models[name].predict_proba(X)[:, 1]
    svm_p = svm_models[name].predict_proba(svm_scalers[name].transform(X))[:, 1]
    nb_p  = nb_models[name].predict_proba(nb_scalers[name].transform(X))[:, 1]
    return np.column_stack([rf_p, svm_p, nb_p])

X_meta_train = build_meta_features(X_train)
X_meta_test  = build_meta_features(X_test)

# --- Subsample for speed (KernelExplainer is slow) ---
np.random.seed(42)
bg_idx     = np.random.choice(len(X_meta_train), 100, replace=False)
background = X_meta_train[bg_idx]

N_EXPLAIN = 50
X_explain = X_meta_test[:N_EXPLAIN]

meta_model = meta_models_v2[name]

# --- Time / memory benchmark ---
tracemalloc.start()
t0 = time.perf_counter()

kernel_explainer = shap.KernelExplainer(meta_model.predict_proba, background)
sv_meta_raw = kernel_explainer.shap_values(X_explain, nsamples=200)

elapsed = time.perf_counter() - t0
current, peak = tracemalloc.get_traced_memory()
tracemalloc.stop()

# --- Normalize KernelExplainer output too ---
sv_meta_pos     = shap_to_2d(sv_meta_raw, positive_class=1)
base_val_meta   = shap_base_to_float(kernel_explainer.expected_value, positive_class=1)
meta_features   = ["RF_prob", "SVM_prob", "NB_prob"]

print(f"⏱ KernelExplainer time: {elapsed:.2f} s  |  "
      f"🧠 Peak memory: {peak / 1024 / 1024:.2f} MiB")
print(f"[{name}] Meta SHAP array shape = {sv_meta_pos.shape}")
assert sv_meta_pos.ndim == 2 and sv_meta_pos.shape[1] == 3

shap_results[name].update({
    "explainer_meta":     kernel_explainer,
    "shap_values_meta":   sv_meta_pos,
    "X_explain_meta":     X_explain,
    "meta_feature_names": meta_features,
    "base_value_meta":    base_val_meta,
})

# --- Summary (beeswarm) on meta-features ---
plt.figure()
shap.summary_plot(
    sv_meta_pos, X_explain,
    feature_names=meta_features,
    show=False, plot_type="dot"
)
plt.title(f"{name} — SHAP Summary (Meta-Model)", fontsize=12, fontweight="bold")
plt.tight_layout()
plt.savefig(f"shap_summary_{name.lower()}_meta.png", dpi=150, bbox_inches="tight")
plt.show()

# --- Bar plot (mean |SHAP|) ---
plt.figure()
shap.summary_plot(
    sv_meta_pos, X_explain,
    feature_names=meta_features,
    show=False, plot_type="bar"
)
plt.title(f"{name} — Mean |SHAP| (Meta-Model)", fontsize=12, fontweight="bold")
plt.tight_layout()
plt.savefig(f"shap_bar_{name.lower()}_meta.png", dpi=150, bbox_inches="tight")
plt.show()

# --- Force plot across all explained samples ---
try:
    shap.force_plot(
        base_val_meta,
        sv_meta_pos,
        X_explain,
        feature_names=meta_features,
        matplotlib=True,
        show=False
    )
    plt.title(f"{name} — SHAP Force Plot (Meta-Model, all explained samples)",
              fontsize=12, fontweight="bold")
    plt.tight_layout()
    plt.savefig(f"shap_force_{name.lower()}_meta.png", dpi=150, bbox_inches="tight")
    plt.show()
except Exception as e:
    print(f"(Force plot skipped: {e})")


# ============================================================
# PART C — Waterfall on the Meta-Model for the same sample
# ============================================================
sv_meta_row = np.asarray(sv_meta_pos[sample_idx if sample_idx < len(sv_meta_pos) else 0]).ravel()
assert sv_meta_row.shape == (3,)

explanation_meta = shap.Explanation(
    values        = sv_meta_row,
    base_values   = base_val_meta,
    data          = X_explain[sample_idx if sample_idx < len(X_explain) else 0].ravel(),
    feature_names = meta_features,
)

plt.figure()
shap.plots.waterfall(explanation_meta, max_display=3, show=False)
plt.title(f"{name} — SHAP Waterfall (Meta-Model, sample #{sample_idx})",
          fontsize=12, fontweight="bold")
plt.tight_layout()
plt.savefig(f"shap_waterfall_{name.lower()}_meta.png", dpi=150, bbox_inches="tight")
plt.show()

print(f"\n✅ Child SHAP analysis (RF + Meta) complete.")
```

```python
# ============================================================
# LIME — Unified explanation for all datasets (RF & Meta)
# ============================================================

import lime
import lime.lime_tabular
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

lime_results = {}

# Helper to build meta-features for the Meta-Model
def build_meta_features(name, X):
    rf_p  = rf_models[name].predict_proba(X)[:, 1]
    svm_p = svm_models[name].predict_proba(svm_scalers[name].transform(X))[:, 1]
    nb_p  = nb_models[name].predict_proba(nb_scalers[name].transform(X))[:, 1]
    return np.column_stack([rf_p, svm_p, nb_p])

for name in ["Child", "Adolescent", "Adult"]:
    print("\n" + "=" * 60)
    print(f"LIME — Dataset: {name}")
    print("=" * 60)

    X_train = splits[name]["X_train"]
    X_test  = locked_test[name]["X_test"]
    y_test  = locked_test[name]["y_test"].values
    y_pred  = rf_results[name]["y_pred"]
    
    # Find the first correctly predicted YES patient
    tp_idx = np.where((y_test == 1) & (y_pred == 1))[0]
    sample_idx = tp_idx[0] if len(tp_idx) > 0 else 0

    # --------------------------------------------------------
    # 1. LIME on Random Forest (All Datasets)
    # --------------------------------------------------------
    print(f"\n[1/2] LIME — {name} : Random Forest")
    
    explainer_rf = lime.lime_tabular.LimeTabularExplainer(
        training_data=np.array(X_train),
        feature_names=list(X_train.columns),
        class_names=["NO", "YES"],
        mode="classification",
        discretize_continuous=True,
        random_state=42
    )

    exp_rf = explainer_rf.explain_instance(
        data_row=X_test.iloc[sample_idx].values,
        predict_fn=rf_models[name].predict_proba,
        num_features=12,
        num_samples=5000,
        top_labels=1
    )
    lime_results[f"{name}_rf"] = exp_rf

    fig = exp_rf.as_pyplot_figure(label=exp_rf.top_labels[0])
    plt.title(f"{name} — LIME (Random Forest, Sample #{sample_idx})", fontweight="bold")
    plt.tight_layout()
    plt.savefig(f"lime_{name.lower()}_rf.png", dpi=150, bbox_inches="tight")
    plt.show()

    # --------------------------------------------------------
    # 2. LIME on Meta-Model (Only Child, per your strategy)
    # --------------------------------------------------------
    if name == "Child":
        print(f"\n[2/2] LIME — {name} : Meta-Model")
        
        X_meta_train = build_meta_features(name, X_train)
        X_meta_test  = build_meta_features(name, X_test)
        meta_model   = meta_models_v2[name]

        explainer_meta = lime.lime_tabular.LimeTabularExplainer(
            training_data=X_meta_train,
            feature_names=["RF_prob", "SVM_prob", "NB_prob"],
            class_names=["NO", "YES"],
            mode="classification",
            discretize_continuous=True,
            random_state=42
        )

        exp_meta = explainer_meta.explain_instance(
            data_row=X_meta_test[sample_idx],
            predict_fn=meta_model.predict_proba,
            num_features=3,
            num_samples=5000,
            top_labels=1
        )
        lime_results[f"{name}_meta"] = exp_meta

        fig = exp_meta.as_pyplot_figure(label=exp_meta.top_labels[0])
        plt.title(f"{name} — LIME (Meta-Model, Sample #{sample_idx})", fontweight="bold")
        plt.tight_layout()
        plt.savefig(f"lime_{name.lower()}_meta.png", dpi=150, bbox_inches="tight")
        plt.show()

print("\n✅ LIME explanations completed for all datasets!")
```
